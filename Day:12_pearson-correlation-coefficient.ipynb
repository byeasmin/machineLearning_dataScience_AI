# Pearson correlation is the most popular correlation measurment technique.

# Correlation with python

```python
import pandas as pd
```

```python
data = {
    'x': [11, 12, 13, 14, 15, 17, 20],
    'y': [12, 13, 15, 14, 16, 16, 18]
}
```

```python
data
```

```python
df = pd.DataFrame(data)
```

```python
df
```

```python
df.head()
```

# Calculate pearson correlation

```python
 corr = df['x'].corr(df['y'])
```

```python
print("pearson correlation coefficient: ", corr)
```
