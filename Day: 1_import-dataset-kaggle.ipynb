## 1. Kaggle Notebook: Adding a Dataset

**In Kaggle, datasets are normally added using the Add Data option.**

**After adding the dataset, Kaggle makes it available under:**

``` text
/kaggle/input/
```

**For example, suppose the dataset is named:**

``` text
my-dataset
```

**The path will generally be:**
``` text
/kaggle/input/my-dataset/
```

## 2. Kaggle: Image Dataset

``` text
/kaggle/input/my-dataset/
│
├── Training/
│   ├── Class_A/
│   ├── Class_B/
│   └── Class_C/
│
└── Testing/
    ├── Class_A/
    ├── Class_B/
    └── Class_C/

```

````python
TRAIN_DIR = "/kaggle/input/datasets/yeasminbenin/brain-tumor-mri/Training"
TEST_DIR = "/kaggle/input/datasets/yeasminbenin/brain-tumor-mri/Testing"
````

````python
from torchvision import datasets

train_dataset = datasets.ImageFolder(TRAIN_DIR)
test_dataset = datasets.ImageFolder(TEST_DIR)
````

## 3. Kaggle: ZIP Dataset

**Sometimes a dataset is provided as a ZIP file.**

``` text
/kaggle/input/my-dataset/data.zip
```

````python
import zipfile

zip_path = "/kaggle/input/my-dataset/data.zip"

with zipfile.ZipFile(zip_path, "r") as zip_ref:
    zip_ref.extractall("/kaggle/working/data")
````
