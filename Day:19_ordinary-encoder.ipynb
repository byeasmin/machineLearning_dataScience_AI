## Data that has a meaningful order is called ordinal data, while data that does not have any meaningful order is called nominal data. When we work with ordinal data, we can use an Ordinal Encoder. Using domain knowledge, we know the exact order of the categories, such as low, medium, and high. We assign numbers according to this meaningful order, so the model can understand the relationship between the categories. That is why Ordinal Encoding is useful for ordinal data.

# **If we apply an Ordinal Encoder to nominal data, then it will not be more effective because nominal data is not suitable for an Ordinal Encoder.**

```python
import pandas as pd
```

```python
from sklearn.preprocessing import OrdinalEncoder
```

```python
df = ['good', 'bad', 'excellent', 'average', 'good', 'average', 'bad', 'excellent', 'good', 'average']
```

```python
data = pd.DataFrame(data = df, columns = ['comparison'])
```

```python
data
```

```python
data.head()
```

# this is the order actually

```python
categories = [['bad', 'average', 'good', 'excellent']]
```

```python
categories
```

# for more details, click shift + tab

```python
encoder = OrdinalEncoder(categories = categories)
```

```python
encoded_data = encoder.fit_transform(data)
encoded_data
```

```python
encoded_data
```

```python
decoded_data = encoder.inverse_transform(encoded_data)
```

```python
decoded_data
```
