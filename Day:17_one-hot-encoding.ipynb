# one hot encoding with python

```python
 import pandas as pd
```

```python
data = {'category': ['a', 'b', 'c', 'a', 'b', 'c','a', 'b', 'c' ]}
```

```python
df = pd.DataFrame(data)
```

```python
df.head()
```

```python
one_hot_en = pd. get_dummies(df, columns = ['category'])
one_hot_en
```

# hover over a default keyword/variable, Press shift + tab the to explore more about encoding or etc variable

```python
one_hot_en = pd. get_dummies(df, columns = ['category'], prefix='Dummy')
one_hot_en
```

```python
one_hot_en = pd. get_dummies(df, columns = ['category'], prefix='encoded')
one_hot_en
```

# hover over prefix to learn more about it

```python
one_hot_en = pd. get_dummies(df, columns = ['category'], prefix='encoded', drop_first= False)
one_hot_en
```

```python
one_hot_en = pd. get_dummies(df, columns = ['category'], prefix='encoded', drop_first= True)
one_hot_en
```
