# R² Score (R-squared)

## **It tells us how well the model explains the variation in the target data.**



## **Its value is usually interpreted like this:**

```` text
1.0 → Perfect prediction
0.90 → Very good fit
0.70 → Explains about 70% of the variation
0.0 → No better than simply predicting the mean
Below 0 → Worse than predicting the mean
