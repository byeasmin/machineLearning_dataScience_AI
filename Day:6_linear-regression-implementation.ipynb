``` text

Y-axis: Exam Score
```

# Linear Regression

## We have to import it from scikit learn, so first of all we have to go to sklearn.org website and then import linear regression



**Their is a lot of regression models**



````python
from sklearn.linear_model import LinearRegression #(press tab to explore more models)
````

# Create object

````python
reg = LinearRegression()
````

````python
df
````

````python
x
````

# Mean with previous data

````python
plt.figure(figsize = (5, 5))
plt.scatter(df['Starting(USD)'], df['Ending(USD)'], color='green')
plt.xlabel('Starting(USD)')
plt.ylabel('Ending(USD)')
plt.title("Stock Prices")
plt.scatter(x.mean(), y.mean(), color='red')
plt.show()
````

# fit, that means training

````python
reg.fit(x, y)
````

````python
m = reg.coef_
````

````python
m
````

````python
c = reg.intercept_
````

````python
c
````

# y = mx + c (manual process)

````python
 m * 16700 + c
````
