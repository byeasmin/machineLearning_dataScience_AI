# Classification of Data

## 1. Based on Data Types

- **Numerical Data**: Data that consists of numbers and can be used in mathematical calculations.
  - **Integer**: Whole numbers, such as `10`, `25`, and `-5`
  - **Float**: Decimal numbers, such as `3.14`, `25.5`, and `-0.75`

- **Categorical Data**: Data that represents categories or labels.
  - **Binary**: Contains only two categories, such as `Yes/No` or `0/1`
  - **Nominal**: Categories without a natural order, such as `Red`, `Blue`, and `Green`
  - **Ordinal**: Categories with a meaningful order, such as `Low`, `Medium`, and `High`

- **Text Data**: Data consisting of words, sentences, or other textual information, such as `"Hello World"` or customer reviews.

- **Date/Time Data**: Data representing dates, times, or both, such as `2026-08-18` or `10:30 AM`.

## 2. Based on Measurement Scale

- **Nominal Scale**
- **Ordinal Scale**
- **Interval Scale**
- **Ratio Scale**

## 3. Based on Role in Analysis

- **Independent Variable (Feature/Input)**
- **Dependent Variable (Target/Output)**
- **Control Variable**
- **Confounding Variable**

## 4. Based on the Nature of the Data

- **Discrete Data**
- **Continuous Data**
- **Qualitative Data**
- **Quantitative Data**

## 5. Based on Distribution

- **Normal Distribution**
- **Uniform Distribution**
- **Skewed Distribution**
  - **Positively Skewed**
  - **Negatively Skewed**
- **Bimodal Distribution**
- **Multimodal Distribution**
