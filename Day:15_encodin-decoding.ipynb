# Categorial Data:
### **Categorical data is data that represents groups, labels, or categories rather than numerical measurements.**
``` 
| Variable           | Categorical data            |
| ------------------ | --------------------------- |
| Eye color          | Brown, Blue, Green          |
| Gender             | Male, Female, Non-binary    |
| Type of car        | Sedan, SUV, Truck           |
| Marital status     | Single, Married, Divorced   |
| Satisfaction level | Poor, Fair, Good, Excellent | 
  ------------------   ---------------------------

```
There are two main types:

Nominal data — categories have no natural order.
Example: colors → Red, Blue, Green.
Ordinal data — categories have a meaningful order.
Example: satisfaction → Poor < Fair < Good < Excellent.

# Continuous Data:

### **Continuous data is numerical data that can take any value within a range, including decimal values.**

```
| Variable       | Examples                  |
| -------------- | ------------------------- |
| Height         | 165.5 cm, 172.3 cm        |
| Weight         | 55.7 kg, 68.25 kg         |
| Temperature    | 36.5°C, 37.25°C           |
| Time           | 2.5 seconds, 2.53 seconds |
| Distance       | 10.2 km, 10.25 km         |
| Blood pressure | 120.5 mmHg                |
| --------------   ------------------------- |
