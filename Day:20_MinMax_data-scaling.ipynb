# **MinMax Normalization**

# Here, Robust Scaler is good for outlier.

# Min-max scaling means converting data from its original range into a fixed range, usually from 0 to 1.

# Normally, data is arranged in different ranges of values, but in min-max scaling, we have to convert the values into a range from 0 to 1.

# Feature Transformation is a part of Feature Engineering

# Raw Code

```python
def minmaxscaling(data):
    min_val = min(data)
    max_val = max(data)
    scaled_data = [(x - min_val) / (max_val - min_val) for x in data]  # list comprehensive example
    return scaled_data
```

# Using Sklearn

```python
import pandas as pd
from sklearn.preprocessing import MinMaxScaler
```

```python
data = {'Feature1': [1, 5, 3, 2, 1],
        'Feature2': [6, 7, 8, 9, 10]}
df = pd.DataFrame(data)
```

```python
df.head()
```

```python
scaler = MinMaxScaler()
scaled_data = scaler.fit_transform(df)
scaled_df =  pd.DataFrame(scaled_data, columns = df.columns)
```

```python
print("Original Data Fame: ")
print(df)
print("Scaled Data Frame: ")
print(scaled_df)
```
