## **Their is 3 categories,** 
- **When both training accuracy and testing accuracy are low, the model is underfitted. This means the model could not learn the data properly, so it gives poor results on both the training data and new data.**

- **When the training accuracy is very high, but the testing accuracy is low, the model is called overfitted. This means the model learned the training data too well, but it cannot give good results on new data.**

- **When both the training accuracy and testing accuracy are high and close to each other, the model is called a good fit. This means the model learned the data properly and can also give good results on new data.**
