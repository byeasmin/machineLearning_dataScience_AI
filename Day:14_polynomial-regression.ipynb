# Polynomial Regression for Non-linear (sin curve) data

````python
import os
````

````python
os.listdir("/kaggle/input/datasets/yeasminbenin/sin-data-for-non-linear-regression")
````

````python
df3 = pd.read_csv("/kaggle/input/datasets/yeasminbenin/sin-data-for-non-linear-regression/sin_dataset.csv")
````

````python
df3.head()
````

````python
x3 = df3[['x2']]
y3 = df3[['y2']]
````

````python
x3
````

````python
y3
````

````python
plt.scatter(x3, y3, color='purple', label='Data with Non-Linear(sin) Relationship')
plt.xlabel('x3')
plt.ylabel('y3')
plt.title('Non-Linear(sin) Relationship')
plt.legend()
plt.tight_layout()
plt.show()
````

````python
p = PolynomialFeatures(degree=9)
x3_p = p.fit_transform(x3)
````

````python
x3_p.shape
````

````python
reg_p = LinearRegression()
````

````python
LinearRegression()
````

````python
reg_p.fit(x3_p, y3)
````

````python
reg_p.score(x3_p, y3)
````

````python
plt.scatter(x3, y3, color='green', label='Data with Non-Linear Relationship Sin Curve')
plt.plot(x3, 2 * np.sin(x3), color='red', label='True Non-Linear Relationship')

plt.xlabel('x3')
plt.ylabel('y3')
plt.title('Non-Linear Relationship Sin Curve')
plt.legend()


plt.tight_layout()
plt.show()
````

# Seperate train test

````python
from sklearn.model_selection import train_test_split
````

````python
xtrain, xtest, ytrain, ytest = train_test_split(x3_p, y3, test_size = 0.30, random_state = 32)
````

````python
reg = LinearRegression()
````

````python
reg.fit(xtrain, ytrain)
````

# Testing score

````python
reg.score(xtest, ytest)
````

# Training Score

````python
reg.score(xtrain, ytrain)
````
