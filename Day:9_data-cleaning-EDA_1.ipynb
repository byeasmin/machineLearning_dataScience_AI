# Fix Remote Work

```python
unique_values = df['Remote_Work'].unique()
print(unique_values)
```

```python
def standardize_boolean(value):
    value = str(value).lower()

    if value in ['yes', 'y', 'true', '1']:
        return True
    elif value in ['no', 'n', 'false', '0']:
        return False
    else:
        return None
```

```python
df['Remote_Work'] = df['Remote_Work'].apply(standardize_boolean)
```
